### C1. Draw a scatter plot between `Runs` and `Price`, and color the points by `Age_Group`.

```python
# C1: Scatter plot with Age_Group coloring

sns.scatterplot(data=data,x="Runs",y="Price",hue="Age_Group")
plt.title(" Scatter Plot of Runs and  Price by Age Group")
plt.xlabel("Runs")
plt.ylabel("Price")
plt.show()
```

Scatter plot uses dots to represent the relationship between two different numeric variables. In this scatter plot the two variables is Runs and Price.

### C2. Create a correlation heatmap using only the columns `Runs`, `Wickets`, `Price`, and `Age`.

```python
# C2: Correlation heatmap
corr = data[["Runs", "Wickets", "Price", "Age"]].corr()

sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

Correlation heatmap is a 2-dimensional graphical representation that uses colored cells to show the correlation coefficients between multiple variables in a dataset
+1 means variable moving in the same direction,-1 means variable moving in the different direction and 0 means there is no relation between the variables.

### C3. Build a bar chart showing the average `Price` by `Experience_Years` and label the axes clearly.

```python
# C3: Average price by experience years
avg_price = data.groupby("Experience_Years")["Price"].mean()

avg_price.plot(kind="bar")

plt.title("Average Price by Experience Years")
plt.xlabel("Experience Years")
plt.ylabel("Average Price")
plt.xticks(rotation=0)

plt.show()
```

Bar Chart is a visual way to compare category wise  data using rectangular bars and the length of each bar is proportional to the value it represents.This bar chart shows  that player with more experience generaly have higher average price.
